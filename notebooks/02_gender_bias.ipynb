## 2. Exploratory Data Analysis (EDA)

## 3. Data Quality Checks

افحصي القيم الناقصة، النصوص الفارغة، التكرار والتصنيفات المتعارضة للنص نفسه. وثقي القرارات وعدد الصفوف قبل وبعد التنظيف. امنعي انتقال النصوص المكررة أو المرتبطة بين أقسام البيانات. التكرار بين tasks مستقلة لا يحذف تلقائيًا.


## 4. Train / Validation / Test Split

اقتراح للنقاش: 80% تدريب، 10% validation، 10% test، وrandom seed = 42. استخدمي stratification عند ملاءمته؛ إذا توجد مجموعات مصادر أو نصوص مرتبطة، اعتمدي group-aware split. احفظي معرفات الصفوف أو manifest لإعادة إنتاج نفس التقسيم. لا تفتحي test للتجارب.


## 5. Dummy Baseline

مرجع بسيط للتأكد أن النموذج يتعلم أفضل من توقع بلا استخدام للنص. اختاري استراتيجية واضحة مثل most_frequent، وسجلي النتائج على validation. ليس نموذج التطبيق النهائي.


## 6. TF-IDF + Logistic Regression

اقتراح baseline أولي يخضع لتأكيد الفريق ومتطلبات الدليل. استخدمي Pipeline، وfit للـTF-IDF على training فقط أو داخل كل fold. سجلي الإعدادات والوقت والنتائج.


## 7. Cross-Validation

اقتراح: Stratified 5-Fold على training فقط؛ استخدمي CV مناسبة للمجموعات إذا احتاجت البيانات. سجلي متوسط Macro F1 والانحراف المعياري. لا تدخلي test في CV أو tuning.
